```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('') # Enter Dataset path
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Here we will find the median of each category in Review Rating column and replace the missing value 
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# As we can see that our attribute names have capital letters and spaces, so we will convert them to snakecase
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_') # replace spaces with underscores
```

```python
df.columns
```

```python
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'}) 
df.columns
```

```python
# creating a column named age_group
labels = ['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], labels = labels, q = 4)

```

```python
df[['age', 'age_group']].head(10)
```

```python
# creating column named purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Monthly': 30,
    'Yearly': 365,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90,
    'Quarterly': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
df[['purchase_frequency_days']].isnull().sum()
```

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

```python
(df.discount_applied == df.promo_code_used).all()
```

```python
# Means every customer who used promo code has availed discount so we need only one column
df = df.drop('promo_code_used', axis = 1)
```
